### 1. Thông tin  dữ liệu
- **Website:** https://www.formula1.com/en/results/2026/drivers
- **Mục tiêu:** Thu thập bảng xếp hạng các tay đua F1 mùa giải 2026.
- **Công cụ sử dụng:** Thư viện `requests` để tải HTML, `BeautifulSoup` để bóc tách dữ liệu và `pandas` để xử lý & xuất file CSV.
- **Các cột dữ liệu:** Pos (vị trí), Driver (tay đua), Nationality (quốc tịch), Car (đội đua), PTS (điểm số).

### 2. Code

#### 2.1. Khai báo thư viện và cấu hình URL, headers

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

url = "https://www.formula1.com/en/results/2026/drivers"
headers = { "User-Agent": ( "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML," " like Gecko) Chrome/120.0.0.0 Safari/537.36")}
```

#### 2.2. Tải và lấy bảng HTML

```python
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table")
```

#### 2.3. Lấy headers của các cột

```python
headers = []
for th in table.find_all("th"):
    text = th.text.strip()
    if text:
        headers.append(text)
```

#### 2.4. Duyệt qua từng dòng

```python
data = []
rows = table.find_all("tr")[1:] 
for tr in rows:
    cols = tr.find_all("td")
    if cols:
        noidung_row = []
        for td in cols:
            text = td.text.strip() 
            if text != "":
                if len(text) > 3 and text[-3:].isupper() and text[-4].islower():
                    text = text[:-3].strip() 
                noidung_row.append(text)
        if len(noidung_row) >= len(headers):
            clean_row = noidung_row[: len(headers)]
            data.append(clean_row)
```

#### 2.5. Lưu thành file csv

```python
df = pd.DataFrame(data, columns=headers)
df.to_csv("f1_drivers_2026.csv", index=False, encoding="utf-8-sig")
print("File da luu")
```
